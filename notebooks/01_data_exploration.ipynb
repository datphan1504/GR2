## 2. Phân bố hướng gói tin (Incoming vs Outgoing)
Trực quan hóa phân bố số lượng gói tin đi vào (-1) và đi ra (+1).

```python
# Trực quan hóa đặc trưng
mock_dir = np.random.choice([-1, 1], size=1000, p=[0.7, 0.3])
plt.figure(figsize=(8, 4))
plt.hist(mock_dir, bins=[-1.5, -0.5, 0.5, 1.5], rwidth=0.8, color='royalblue')
plt.xticks([-1, 1], ['Incoming (-1)', 'Outgoing (+1)'])
plt.title('Packet Direction Distribution')
plt.ylabel('Count')
plt.grid(True, alpha=0.3)
plt.show()
```
